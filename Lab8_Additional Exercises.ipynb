```python
# Exercise 1

class Book:
    def __init__(self, title, author, available=True):
        self.title = title
        self.author = author
        self.available = available

class Library:
    def __init__(self):
        self.books = []

    def add_books(self, n):
        for _ in range(n):
            title = input("Enter the book title: ").strip()
            author = input("Enter the book author: ").strip()
            book = Book(title, author)
            self.books.append(Book(title, author, True))
            print(f"Added '{title}' by {author} to the library.")

    def view_books(self):
        print("\n---Library Books---")
        for book in self.books:
            status = "Available" if book.available else "Borrowed"
            print(f"{book.title} by {book.author} - {status}")

    def borrow_book(self, title):
        for book in self.books:
            if book.title == title and book.available:
                book.available = False
                print(f"You have borrowed '{book.title}'.")
                return
        else:
            print(f"Sorry, '{title}' is not available for borrowing.")
            return
        print(f"Sorry, '{title}' is not recognized by the library.")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if not book.available:
                    book.available = True
                    print(f"You have returned '{book.title}'.")
                else:
                    print(f"'{book.title}' was not borrowed.")
                return
        print(f"Sorry, '{title}' is not recognized by the library.")

lib = Library()

n = int(input("How many books to add? "))
lib.add_books(n)

while True:
    print("\n1. View Books\n2. Borrow Book\n3. Return Book\n4. Exit")
    choice = input("Choose: ")

    if choice == "1":
        lib.view_books()
    elif choice == "2":
        lib.borrow_book(input("Enter title to borrow: "))
    elif choice == "3":
        lib.return_book(input("Enter title to return: "))
    elif choice == "4":
        break
```

```python
# Exercise 2

class Employee:
    def __init__(self, name, rate, hours):
        self.name = name
        self.rate = rate
        self.hours = hours

    def calculate_salary(self):
        if self.hours <= 160:
            salary = self.rate * self.hours
            overtime_pay = 0
            overall_salary = salary
        else:
            overtime_hours = self.hours - 160
            overtime_pay = overtime_hours * self.rate * 1.5
            salary = self.rate * 160
            overall_salary = (self.rate * 160) + overtime_pay

        return salary, overtime_pay, overall_salary

employees = []
n = int(input("Enter number of employees: "))

if n < 1:
    print("Number of employees must be at least 1.")
else:
    for _ in range(n):
        name = input("Enter employee name: ").strip()
        rate = float(input("Enter hourly rate: "))
        hours = float(input("Enter hours worked: "))

        if rate < 0 or hours < 0:
            print("Rate and hours must be non-negative.")
            continue

        employees.append(Employee(name, rate, hours))

    print("\n---Employee Salaries---")
    for emp in employees:
        salary, overtime_pay, overall_salary = emp.calculate_salary()
        print(f"Employee's Name: {emp.name}")
        print(f"Total Hours Worked: {emp.hours}")
        print(f"Base Salary: {salary}")
        print(f"Overtime Pay: {overtime_pay}")
        print(f"Overall Salary: {overall_salary}\n")
```

```python
# Exercise 3

class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

class Cart:
    def __init__(self):
        self.items = {}
        self.total = 0

products = [
    Product("Hair Clamp", 50, 20),
    Product("Flex Stick Ballpen", 8, 60),
    Product("Water Bottle", 500, 10),
    Product("Notebook", 25, 20),
    Product("ID Lace", 55, 85)
]

cart = Cart()

def show_products():
    print("\n--- Products ---")
    for i, p in enumerate(products):
        print(f"{i+1}. {p.name} - ₱{p.price} (Stock: {p.stock})")

def find_product(name):
    for p in products:
        if p.name.lower() == name.lower():
            return p
    return None

while True:
    print("\n1. View Products\n2. Add to Cart\n3. Remove from Cart\n4. View Cart\n5. Exit")
    choice = input("Choose: ")

    if choice == "1":
        show_products()

    elif choice == "2":
        show_products()
        name = input("Enter product name: ")
        qty = input("Enter quantity: ")

        if not qty.isdigit():
            print("Invalid quantity input.")
            continue

        qty = int(qty)
        product = find_product(name)

        if product:
            if qty <= product.stock:
                product.stock -= qty
                cart.items[name] = cart.items.get(name, 0) + qty
                cart.total += product.price * qty
                print(f"Added {qty} {product.name}(s) to cart.")
            else:
                print(f"Not enough stock of {product.name} available.")
        else:
            print("Invalid product selection.")

    elif choice == "3":
        name = input("Enter product name to remove: ")
        product = find_product(name)

        if product and name in cart.items:
            qty = cart.items[name]

            product.stock += qty
            cart.total -= product.price * qty

            del cart.items[name]
            print(f"Removed {name} from cart.")
        else:
            print("Item not found in cart.")

    # ---------------- VIEW CART ----------------
    elif choice == "4":
        print("\n--- CART CONTENTS ---")

        if not cart.items:
            print("Cart is empty.")
        else:
            for name, qty in cart.items.items():
                print(f"{name} x{qty}")
            print(f"Total Price: ₱{cart.total}")

    # ---------------- EXIT ----------------
    elif choice == "5":
        print("Checkout complete. Thank you!")
        break

    else:
        print("Invalid option. Try again.")
```
